#**Exploratory Data Analysis**

```python
# Analyze the fuel type distribution

plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Fuel_Type")

plt.title("Fuel Type Distribution")
plt.xlabel("Fuel Type")
plt.ylabel("Count")

plt.show()
```

```python
# Visualize the fuel type distribution

plt.figure(figsize=(6,6))

plt.pie(
    df["Fuel_Type"].value_counts(),
    labels=df["Fuel_Type"].value_counts().index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Fuel Type Distribution")

plt.show()
```

```python
# Analyze the distribution of seller types

plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Seller_Type')
plt.title('Seller Type Distribution')
plt.xlabel('Seller Type')
plt.ylabel('Count')
plt.show()
```

```python
# Analyze the distribution of transmission types

plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Transmission')
plt.title('Transmission Distribution')
plt.xlabel('Transmission')
plt.ylabel('Count')
plt.show()
```

```python
# Analyze the distribution of vehicle selling prices

plt.figure(figsize=(7,4))
sns.histplot(df['Selling_Price'], bins=20, kde=True)
plt.title('Selling Price Distribution')
plt.xlabel('Selling Price (Lakhs)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Analyze the distribution of vehicle manufacturing years

plt.figure(figsize=(10,4))
sns.countplot(data=df, x='Year')
plt.xticks(rotation=90)
plt.title('Manufacturing Year Distribution')
plt.xlabel('Manufacturing Year')
plt.ylabel('Count')
plt.show()
```

```python
# Identify the most frequently listed vehicle model

df["Car_Name"].value_counts().head(10)
```

```python
# Visualize the top 10 most frequently listed vehicle models

plt.figure(figsize=(10,5))
df["Car_Name"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Most Frequently Listed Vehicle Models")
plt.xlabel("Vehicle Model")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Check whether the dataset contains CNG vehicles

cng = df[df["Fuel_Type"]=="CNG"]

print("Total CNG Vehicles :", len(cng))

cng
```

```python
# Count vehicles sold directly by individual owners

individual = df[df["Seller_Type"]=="Individual"]

print("Vehicles Sold by Individuals :", len(individual))
```

```python
# Check the number of automatic transmission vehicles

automatic = df[df["Transmission"]=="Automatic"]

print("Automatic Vehicles :", len(automatic))
```

```python
# Count single-owner vehicles

single_owner = df[df["Owner"]==0]

print("Single Owner Vehicles :", len(single_owner))
```

```python
# Calculate vehicle depreciation

df["Depreciation"] = df["Present_Price"] - df["Selling_Price"]

df[["Car_Name","Present_Price","Selling_Price","Depreciation"]].head()
```

```python
# Identify the vehicle with the highest depreciation

df.sort_values("Depreciation", ascending=False).head(1)
```

```python
# Identify the vehicle with the lowest depreciation

df.sort_values("Depreciation").head(1)
```

```python
# Analyze average depreciation by vehicle model

brand_dep = df.groupby("Car_Name")["Depreciation"].mean().sort_values()

brand_dep.head(10)
```

```python
# Visualize the vehicle models with the lowest average depreciation

plt.figure(figsize=(10,5))
brand_dep.head(10).plot(kind="barh")
plt.title("Vehicle Models with Lowest Average Depreciation")
plt.xlabel("Average Depreciation")
plt.ylabel("Vehicle Model")
plt.show()
```

# **Advanced Vehicle Analysis**

```python
# Analyze vehicles manufactured after 2014

new_vehicles = df[df["Year"] > 2014]

new_vehicles
```

```python
# Extract only two-wheelers from the dataset

bikes = df[df["Car_Name"].str.contains(
    "Activa|Honda|Royal|Bajaj|TVS|Hero|Yamaha|KTM|Suzuki",
    case=False,
    na=False
)]

print("Total Bikes :", len(bikes))

bikes.head()
```

```python
# Identify the oldest bike in the dataset

bikes.sort_values("Year").head(1)
```

```python
# Identify the newest bike in the dataset

bikes.sort_values("Year", ascending=False).head(1)
```

```python
# Identify the most frequently listed bike model

bikes["Car_Name"].value_counts()
```

```python
# Visualize the top 10 most frequently listed bike models

plt.figure(figsize=(10,5))

bikes["Car_Name"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Most Frequently Listed Bike Models")
plt.xlabel("Bike Model")
plt.ylabel("Count")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
# Analyze the highest selling two-wheelers

bikes.sort_values("Selling_Price", ascending=False)[
    ["Car_Name","Year","Selling_Price","Present_Price","Kms_Driven"]
].head()
```

```python
# Extract only cars from the dataset

cars = df.drop(bikes.index)

print("Total Cars :", len(cars))

cars.head()
```

```python
# Identify the oldest car in the dataset

cars.sort_values("Year").head(1)
```

```python
# Identify the newest car in the dataset

cars.sort_values("Year", ascending=False).head(1)
```

```python
# Analyze the highest selling cars

cars.sort_values("Selling_Price", ascending=False)[
    ["Car_Name","Year","Selling_Price","Present_Price","Kms_Driven"]
].head()
```

```python
# Analyze the relationship between manufacturing year and selling price

plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x="Year", y="Selling_Price")

plt.title("Manufacturing Year vs Selling Price")
plt.xlabel("Manufacturing Year")
plt.ylabel("Selling Price (Lakhs)")

plt.show()
```

```python
# Analyze the relationship between kilometers driven and selling price

plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x="Kms_Driven", y="Selling_Price")

plt.title("Kilometers Driven vs Selling Price")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (Lakhs)")

plt.show()
```

# **Conclusion**

This analysis of the CarDekho dataset provided valuable insights into the used vehicle market.

**Key Findings**

- Vehicles manufactured between 2003 and 2018 are present in the dataset.
- Petrol vehicles are the most common.
- Manual transmission vehicles dominate the market.
- Most vehicles belong to the first owner.
- Newer vehicles generally have higher resale values.
- Vehicle age and kilometers driven significantly influence selling price.
- Premium vehicles experience higher depreciation but retain strong resale value.
